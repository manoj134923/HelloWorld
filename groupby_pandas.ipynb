```python
import  pandas as pd
data={
    "Name":["Manoj","Ravi","Neha","Priya"],
   "Department":["Patna","Rajkot","Mumbai","Delhi"],
    "Marks":[98,90,95,94]
    
}
df=pd.DataFrame(data)
print(df)
```

```python
# groupby important method 
# 1.sum(),2.max(),3.min(),4.mean(),5.count(),6.agg(),7,size()
```

```python
Sum_marks=df.groupby("Name")["Marks"].sum()
print(Sum_marks)
```

```python
Sum_mean=df.groupby("Name")["Marks"].mean()
print(Sum_mean)
```

```python
data_size=df.groupby("Name")["Marks"].size()
print(data_size)
```

```python
agg_marks=df.groupby("Department")["Marks"].agg(["sum","max","count","mean"])
print(agg_marks)
```

```python
# groupby multiple column 
```

```python
data=df.groupby(["Department","Name"])["Marks"].sum()
print(data)
```
